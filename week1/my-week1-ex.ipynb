```python
MODEL_GPT = 'gpt-4o-mini'
MODEL_LLAMA = 'llama3.2'
```

```python
import os
from dotenv import load_dotenv
from IPython.display import Markdown, display, update_display
from openai import OpenAI
```

```python
system_prompt = "You are an expert Python developer and should be able to answer any Python programming construct \
  in very simple way that even 10 year old could understand, then give simplest programming example in python on how to use it. Respond in Markdown format. Python code will be given to you in the following format\n"
system_prompt += """Please explain what this code does and why:
  <code>
    yield from {book.get("author") for book in books if book.get("author")}
  </code>"
  """
```

```python
load_dotenv(override=True)
openai = OpenAI()
```

```python
def get_user_prompt(code_construct):
  user_prompt = f"""Please explain what this code does and why:
<code>
  {code_construct}
</code>"""
  return user_prompt
```

```python
get_user_prompt("if x > 0:")
```

```python
def get_gpt_code_explanation(code_construct):
  stream = openai.chat.completions.create(
    model=MODEL_GPT,
    messages=[
      { "role": "system", "content" : system_prompt },
      { "role": "user", "content": get_user_prompt(code_construct) }
    ],
    stream=True
  )

  response = ""
  display_handle = display(Markdown(""), display_id=True)
  for chunk in stream:
    response += chunk.choices[0].delta.content or ''
    update_display(Markdown(response), display_id=display_handle.display_id)
```

```python
get_gpt_code_explanation("yield from {book.get('author') for book in books if book.get('author')}")
```
